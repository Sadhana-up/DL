```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

# 

```python
np.random.seed(42)
X_train = np.random.uniform(0, 5, 150).reshape(-1, 1)
y_train = 3 * X_train + np.random.randn(150, 1) * 5
```

```python
X_test = np.random.uniform(5, 10, 100).reshape(-1, 1)
y_test = 3 * X_test + np.random.randn(100, 1) * 5
```

```python
plt.scatter(X_train, y_train, label="Train Data")
plt.scatter(X_test, y_test, label="Test Data")
plt.legend()
plt.title("Before: Raw Linear Data with Noise")
plt.show()
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
X_all = np.vstack((X_train, X_test))
y_pred_all = model.predict(X_all)
```

```python
plt.scatter(X_train, y_train, label="Train Data")
plt.scatter(X_test, y_test, label="Test Data")
plt.plot(X_all, y_pred_all, color='red')
plt.legend()
plt.title("After: Linear Regression")
plt.show()
```

```python
def print_metrics(y_true, y_pred, name):
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    print(f"{name} -> R2: {r2:.4f}, MSE: {mse:.2f}, RMSE: {rmse:.2f}")

print("\nOVERFITTING CASE")
print_metrics(y_train, model.predict(X_train), "Train")
print_metrics(y_test, model.predict(X_test), "Test")
```
