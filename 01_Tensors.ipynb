### Creating a Tensor

```python
# using empty
a = torch.empty(2,3) ##shape
```

```python
# type of tensor
type(a)
```

```python
# using zeros
torch.zeros(2,3)
```

```python
#using rand -> value from 0 to 1 randomly
torch.rand(2,3)
```

```python
#using tensor
torch.tensor([[1,2,3],[4,5,6]])
```
