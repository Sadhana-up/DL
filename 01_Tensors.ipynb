Reductiom Operation

```python
e = torch.randint(size = (2,3),low=0,high=10)
```

```python
# sum
torch.sum(e)
#along cols
torch.sum(e,dim=0)
#along rows
torch.sum(e,dim=1)
```

```python
e
```
